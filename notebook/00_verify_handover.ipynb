```python
import sqlite3
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "data" / "raw").exists():
    for candidate in [PROJECT_ROOT.parent, PROJECT_ROOT / "notebook"]:
        if (candidate / "data" / "raw").exists():
            PROJECT_ROOT = candidate
            break
DB_PATH = PROJECT_ROOT / "data" / "raw" / "airtel.db"

con = sqlite3.connect(DB_PATH)

def q(sql):
    return pd.read_sql(sql, con)
```

```python
q("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name;
""")
```

```python
q("""
SELECT 'circles' AS tbl, COUNT(*) AS n FROM circles
UNION ALL
SELECT 'plans', COUNT(*) FROM plans
UNION ALL
SELECT 'customers', COUNT(*) FROM customers
UNION ALL
SELECT 'usage_daily', COUNT(*) FROM usage_daily
UNION ALL
SELECT 'recharges', COUNT(*) FROM recharges
UNION ALL
SELECT 'complaints', COUNT(*) FROM complaints
ORDER BY tbl;
""")
```
